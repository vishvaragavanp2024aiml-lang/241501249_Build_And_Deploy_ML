```python
register_html = """
<!DOCTYPE html>
<html>
<head>
<title>Register</title>

<style>
body {
    background-color: greenyellow;
    font-family: cursive;
}

.box {
    width: 400px;
    margin: 80px auto;
    padding: 40px;
    background-color: aqua;
    border-radius: 10px;
    box-shadow: 0 0 15px gray;
}

input {
    width: 95%;
    padding: 12px;
    margin: 10px 0;
}

button {
    width: 100%;
    padding: 12px;
    background-color: green;
    color: white;
    border: none;
    font-size: 16px;
}
</style>
</head>

<body>

<div class="box">

<h2 style="text-align:center;">Register</h2>

<form method="POST">

<input type="text" name="username"
placeholder="Enter your name" required>

<input type="password" name="password"
placeholder="Enter password" required>

<button type="submit">Register</button>

</form>

<p style="text-align:center;">
Already have an account?
<a href="/login">Sign In</a>
</p>

</div>

</body>
</html>
"""

login_html = """
<!DOCTYPE html>
<html>
<head>
<title>Login</title>

<style>
body {
    background-color: greenyellow;
    font-family: cursive;
}

.box {
    width: 400px;
    margin: 80px auto;
    padding: 40px;
    background-color: aqua;
    border-radius: 10px;
    box-shadow: 0 0 15px gray;
}

input {
    width: 95%;
    padding: 12px;
    margin: 10px 0;
}

button {
    width: 100%;
    padding: 12px;
    background-color: green;
    color: white;
    border: none;
    font-size: 16px;
}
</style>
</head>

<body>

<div class="box">

<h2 style="text-align:center;">Login</h2>

<form method="POST">

<input type="text" name="username"
placeholder="Enter your username" required>

<input type="password" name="password"
placeholder="Enter your password" required>

<button type="submit">Login</button>

</form>

<p style="text-align:center;">
Don't have an account?
<a href="/register">Register</a>
</p>

</div>

</body>
</html>
"""

@app.route("/register", methods=["GET", "POST"])
def register():

    if request.method == "POST":

        username = request.form["username"]
        password = request.form["password"]

        users[username] = password

        return redirect(url_for("login"))

    return render_template_string(register_html)


@app.route("/login", methods=["GET", "POST"])
def login():

    if request.method == "POST":

        username = request.form["username"]
        password = request.form["password"]

        if username in users and users[username] == password:

            session["username"] = username

            return redirect(url_for("dashboard"))

        return "Invalid username or password"

    return render_template_string(login_html)


print("Register and Login pages created successfully.")
```

```python
predictor_html = """
<!DOCTYPE html>
<html>
<head>

<title>Stock Price Predictor</title>

<style>

body {
    background-color: greenyellow;
    font-family: cursive;
}

.box {
    width: 450px;
    margin: 60px auto;
    padding: 35px;
    background-color: aqua;
    border-radius: 10px;
    box-shadow: 0 0 15px gray;
}

h1 {
    color: black;
    text-align: center;
}

input {
    width: 95%;
    padding: 12px;
    margin: 8px 0;
}

button {
    width: 100%;
    padding: 12px;
    background-color: green;
    color: white;
    border: none;
    font-size: 16px;
}

.result {
    background-color: white;
    color: black;
    padding: 15px;
    margin-top: 20px;
    text-align: center;
    font-size: 20px;
}

.logout {
    text-align: center;
    margin-top: 20px;
}

</style>

</head>

<body>

<div class="box">

<h1>Stock Price Predictor</h1>

<h2>Welcome, {{ username }}!</h2>

<form method="POST">

<input type="number"
name="Open"
placeholder="Opening Price"
step="any"
required>

<input type="number"
name="High"
placeholder="Highest Price"
step="any"
required>

<input type="number"
name="Low"
placeholder="Lowest Price"
step="any"
required>

<input type="number"
name="Volume"
placeholder="Volume"
required>

<button type="submit">Predict</button>

</form>

{% if prediction_text %}

<div class="result">
{{ prediction_text }}
</div>

{% endif %}

<div class="logout">
<a href="/logout">Logout</a>
</div>

</div>

</body>
</html>
"""

@app.route("/dashboard", methods=["GET", "POST"])
def dashboard():

    if "username" not in session:
        return redirect(url_for("login"))

    prediction_text = ""

    if request.method == "POST":

        open_price = float(request.form["Open"])
        high = float(request.form["High"])
        low = float(request.form["Low"])

        # Prediction logic used in the lab manual
        predicted_price = (
            open_price + high + low
        ) / 3

        prediction_text = (
            f"Predicted Stock Price: ₹{predicted_price:.2f}"
        )

    return render_template_string(
        predictor_html,
        username=session["username"],
        prediction_text=prediction_text
    )


@app.route("/logout")
def logout():

    session.pop("username", None)

    return redirect(url_for("login"))


print("Stock Price Predictor page created successfully.")
```

```python
import threading
import time
def run_app():
    app.run(
        host="127.0.0.1",
        port=5000,
        debug=False,
        use_reloader=False
    )

thread = threading.Thread(
    target=run_app,
    daemon=True
)
thread.start()
time.sleep(2)
print("==================================================")
print("Stock Price Prediction Application Started")
print("==================================================")
print("Open the following URL in your browser:")
print("http://127.0.0.1:5000")
```
